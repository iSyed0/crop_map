```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.colors import ListedColormap, BoundaryNorm

import xgboost as xgb
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (accuracy_score, cohen_kappa_score,
                             classification_report, confusion_matrix)

import rasterio

print("XGBoost version:", xgb.__version__)
```

```python
# Path to the training CSV (uploaded to the session, or on Drive)
CSV_PATH = "/content/Training_Samples_Balanced (1).csv"
# e.g. CSV_PATH = f"{OUT_DIR}/Training_Samples_Balanced.csv"

df = pd.read_csv(CSV_PATH)
print("Shape:", df.shape)
df.head()
```

```python
# ----- Columns used for modelling (13 features only) -----
FEATURES   = ['B2','B3','B4','B5','B6','B7','B8','B8A','B11','B12','NDVI','NDMI','NDWI']
LABEL_ID   = 'crop_id'      # numeric class (1-12)
LABEL_NAME = 'class_name'   # human-readable label
# NOTE: longitude, latitude, year and source_dataset are intentionally NOT used as features.

# Mapping: class number -> label
class_lookup = (df[[LABEL_ID, LABEL_NAME]]
                .drop_duplicates()
                .sort_values(LABEL_ID))
id_to_name = dict(zip(class_lookup[LABEL_ID], class_lookup[LABEL_NAME]))

print("Classes found:")
for k, v in id_to_name.items():
    print(f"  {k:>2} : {v}")
```

```python
# Class distribution
counts = df[LABEL_NAME].value_counts().sort_index()

plt.figure(figsize=(10, 5))
plt.bar(counts.index, counts.values, color='seagreen')
plt.xticks(rotation=45, ha='right')
plt.ylabel("Number of samples")
plt.title("Training samples per class")
plt.tight_layout()
plt.show()

counts
```

```python
X     = df[FEATURES]
y_raw = df[LABEL_ID]

# Encode class numbers to 0..n-1 for XGBoost
le = LabelEncoder()
y  = le.fit_transform(y_raw)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42, stratify=y
)

print(f"Training samples : {X_train.shape[0]}")
print(f"Testing samples  : {X_test.shape[0]}")
print(f"Number of classes: {len(le.classes_)}")
```

```python
model = xgb.XGBClassifier(
    n_estimators=400,
    max_depth=6,
    learning_rate=0.1,
    subsample=0.8,
    colsample_bytree=0.8,
    objective='multi:softprob',
    num_class=len(le.classes_),
    eval_metric='mlogloss',
    tree_method='hist',
    random_state=42,
    n_jobs=-1,
)

model.fit(X_train, y_train)
print("Model training complete.")
```

```python
y_pred = model.predict(X_test)

# class numbers (in encoded order) and their labels
class_ids   = le.classes_                        # e.g. [1, 2, ..., 12]
class_names = [id_to_name[c] for c in class_ids]

oa    = accuracy_score(y_test, y_pred)
kappa = cohen_kappa_score(y_test, y_pred)

print(f"Overall Accuracy : {oa*100:.2f}%")
print(f"Kappa Coefficient: {kappa:.4f}\n")
print(classification_report(y_test, y_pred, target_names=class_names))
```

```python
# Confusion matrix
cm = confusion_matrix(y_test, y_pred)

fig, ax = plt.subplots(figsize=(9, 8))
im = ax.imshow(cm, cmap='Blues')
ax.set_xticks(range(len(class_names)))
ax.set_yticks(range(len(class_names)))
ax.set_xticklabels(class_names, rotation=45, ha='right')
ax.set_yticklabels(class_names)
ax.set_xlabel("Predicted")
ax.set_ylabel("Actual")
ax.set_title("Confusion Matrix")

thresh = cm.max() / 2.0
for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        ax.text(j, i, cm[i, j], ha='center', va='center',
                color='white' if cm[i, j] > thresh else 'black', fontsize=8)

fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
plt.tight_layout()
plt.show()
```

```python
# ---- 6b.1 Stratified K-Fold Cross-Validation (baseline hyperparameters) ----
# 5-fold StratifiedKFold on the training set only (X_test stays untouched for the
# final hold-out check later). Stratified so each fold keeps the same per-class
# proportions as the full training set -- important here since the 12 crop classes
# are not perfectly balanced (see the Section 3 bar chart).
from sklearn.model_selection import StratifiedKFold, cross_val_score, RandomizedSearchCV

cv_strategy = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

baseline_cv_model = xgb.XGBClassifier(
    n_estimators=400,
    max_depth=6,
    learning_rate=0.1,
    subsample=0.8,
    colsample_bytree=0.8,
    objective='multi:softprob',
    num_class=len(le.classes_),
    eval_metric='mlogloss',
    tree_method='hist',
    random_state=42,
    n_jobs=-1,
)

cv_scores = cross_val_score(
    baseline_cv_model, X_train, y_train,
    cv=cv_strategy, scoring='accuracy', n_jobs=-1,
)

print("Per-fold accuracy :", np.round(cv_scores, 4))
print(f"CV mean accuracy  : {cv_scores.mean()*100:.2f}%")
print(f"CV std deviation  : {cv_scores.std()*100:.2f}%")
print(f"(For comparison, the single hold-out test accuracy in Section 6 was 65.85%.)")
```

```python
# ---- 6b.2 Hyperparameter Tuning (RandomizedSearchCV, 5-fold stratified) ----
# Searches over tree structure (max_depth, min_child_weight, gamma), learning
# rate / number of trees, row & column subsampling, AND explicit L1/L2
# regularization (reg_alpha, reg_lambda) -- the knobs the original Section 5 model
# left at their defaults. Every candidate is scored with the same cv_strategy
# used above, so results are directly comparable to the baseline CV score.
param_grid = {
    'n_estimators':     [200, 300, 400, 600],
    'max_depth':        [4, 5, 6, 8],
    'learning_rate':    [0.01, 0.05, 0.1, 0.2],
    'subsample':        [0.6, 0.7, 0.8, 1.0],
    'colsample_bytree': [0.6, 0.7, 0.8, 1.0],
    'min_child_weight': [1, 3, 5],
    'gamma':            [0, 0.1, 0.3],
    'reg_alpha':        [0, 0.1, 1],
    'reg_lambda':       [1, 2, 5],
}

search_base_model = xgb.XGBClassifier(
    objective='multi:softprob',
    num_class=len(le.classes_),
    eval_metric='mlogloss',
    tree_method='hist',
    random_state=42,
    n_jobs=-1,
)

random_search = RandomizedSearchCV(
    estimator=search_base_model,
    param_distributions=param_grid,
    n_iter=25,               # 25 random combinations out of the full grid
    scoring='accuracy',
    cv=cv_strategy,          # same StratifiedKFold(5) as 6b.1
    random_state=42,
    n_jobs=-1,
    verbose=1,
)
random_search.fit(X_train, y_train)

print(f"\nBest CV accuracy : {random_search.best_score_*100:.2f}%")
print("Best parameters   :")
for k, v in random_search.best_params_.items():
    print(f"  {k:>17}: {v}")

best_params = random_search.best_params_
```

```python
# ---- 6b.3 Refit best model with train/validation loss tracking + early stopping ----
tuned_model = xgb.XGBClassifier(
    **best_params,
    objective='multi:softprob',
    num_class=len(le.classes_),
    eval_metric='mlogloss',
    tree_method='hist',
    random_state=42,
    n_jobs=-1,
    early_stopping_rounds=20,
)

tuned_model.fit(
    X_train, y_train,
    eval_set=[(X_train, y_train), (X_test, y_test)],   # [0]=train, [1]=validation
    verbose=False,
)

print(f"Best iteration (early stopping) : {tuned_model.best_iteration} "
      f"/ {best_params.get('n_estimators')} requested trees")
```

```python
# ---- 6b.4 Train vs. Validation loss curve (the actual "loss difference") ----
results    = tuned_model.evals_result()
train_loss = results['validation_0']['mlogloss']
val_loss   = results['validation_1']['mlogloss']
rounds     = range(1, len(train_loss) + 1)

plt.figure(figsize=(9, 5))
plt.plot(rounds, train_loss, label='Train mlogloss')
plt.plot(rounds, val_loss,   label='Validation (test) mlogloss')
plt.axvline(tuned_model.best_iteration, color='gray', linestyle='--',
            label=f'Best iteration ({tuned_model.best_iteration})')
plt.xlabel('Boosting round')
plt.ylabel('Multiclass log loss')
plt.title('Train vs Validation Loss (overfitting / underfitting check)')
plt.legend()
plt.tight_layout()
plt.show()

final_gap = val_loss[tuned_model.best_iteration] - train_loss[tuned_model.best_iteration]
print(f"Train loss @ best iteration      : {train_loss[tuned_model.best_iteration]:.4f}")
print(f"Validation loss @ best iteration : {val_loss[tuned_model.best_iteration]:.4f}")
print(f"Loss gap (validation - train)    : {final_gap:.4f}")
print("A small, stable gap = good fit. A gap that keeps widening as rounds increase = "
      "overfitting. Both curves flat and high = underfitting.")
```

```python
# ---- 6b.5 Overfitting / Underfitting verdict + final accuracy comparison ----
train_pred = tuned_model.predict(X_train)
test_pred  = tuned_model.predict(X_test)

train_acc        = accuracy_score(y_train, train_pred)
test_acc         = accuracy_score(y_test, test_pred)
test_kappa_tuned = cohen_kappa_score(y_test, test_pred)

print(f"Train accuracy : {train_acc*100:.2f}%")
print(f"Test accuracy  : {test_acc*100:.2f}%")
print(f"Accuracy gap   : {(train_acc-test_acc)*100:.2f} pts")
print(f"Test Kappa     : {test_kappa_tuned:.4f}  (baseline Section 6 Kappa was 0.6242)")

gap = train_acc - test_acc
if gap > 0.15:
    verdict = "OVERFITTING -- large train/test gap, model has memorised training noise."
elif train_acc < 0.75 and test_acc < 0.75:
    verdict = ("Possible UNDERFITTING -- both train and test accuracy are limited; "
               "model/features may be too weak to separate these classes further.")
else:
    verdict = "Reasonable fit -- train/test gap is small and both scores are broadly consistent."
print("\nVerdict:", verdict)

print("\nTuned model classification report (test set):")
print(classification_report(y_test, test_pred, target_names=class_names))

# Downstream cells (Section 9 onward) call `model.predict(...)` -- point that name at the
# tuned + validated model so the Islamabad classification benefits from this section too.
model = tuned_model
print("`model` now refers to the tuned, cross-validated XGBClassifier for all downstream steps.")
```

```python
# ---- 6c.1 Stricter Hyperparameter Search (targets the overfitting found in 6b) ----
# 6b's search only scored candidates on accuracy, so it freely picked the LEAST
# restrictive regularization values (gamma=0, min_child_weight=1, colsample_bytree=1.0)
# because they didn't cost any CV accuracy -- but they also didn't stop the model from
# memorising the training rows (95.66% train vs 65.64% test, loss gap 0.7307). This pass
# scores on 'neg_log_loss' instead (punishes over-confident/overfit predictions, not
# just final class correctness) and removes the loosest candidate values from the grid.
strict_param_grid = {
    'n_estimators':     [150, 200, 300],
    'max_depth':        [3, 4, 5],
    'learning_rate':    [0.01, 0.03, 0.05],
    'subsample':        [0.5, 0.6, 0.7],
    'colsample_bytree': [0.5, 0.6, 0.7],
    'min_child_weight': [5, 10, 20],
    'gamma':            [0.5, 1, 2],
    'reg_alpha':        [1, 2, 5],
    'reg_lambda':       [5, 10, 20],
}

strict_search_base = xgb.XGBClassifier(
    objective='multi:softprob',
    num_class=len(le.classes_),
    eval_metric='mlogloss',
    tree_method='hist',
    random_state=42,
    n_jobs=-1,
)

strict_search = RandomizedSearchCV(
    estimator=strict_search_base,
    param_distributions=strict_param_grid,
    n_iter=25,
    scoring='neg_log_loss',   # <- punishes overfit/overconfident predictions, not just accuracy
    cv=cv_strategy,
    random_state=42,
    n_jobs=-1,
    verbose=1,
)
strict_search.fit(X_train, y_train)

print(f"Best CV log-loss  : {-strict_search.best_score_:.4f}  (lower is better)")
print("Best parameters    :")
for k, v in strict_search.best_params_.items():
    print(f"  {k:>17}: {v}")

strict_best_params = strict_search.best_params_
```

```python
# ---- 6c.2 Refit with tighter early stopping (patience 10 instead of 20) ----
strict_model = xgb.XGBClassifier(
    **strict_best_params,
    objective='multi:softprob',
    num_class=len(le.classes_),
    eval_metric='mlogloss',
    tree_method='hist',
    random_state=42,
    n_jobs=-1,
    early_stopping_rounds=10,
)

strict_model.fit(
    X_train, y_train,
    eval_set=[(X_train, y_train), (X_test, y_test)],
    verbose=False,
)

print(f"Best iteration : {strict_model.best_iteration} / "
      f"{strict_best_params.get('n_estimators')} requested trees")
```

```python
# ---- 6c.3 Loss curve + direct comparison against the 6b (loose) tuned model ----
strict_results    = strict_model.evals_result()
strict_train_loss = strict_results['validation_0']['mlogloss']
strict_val_loss   = strict_results['validation_1']['mlogloss']
strict_rounds     = range(1, len(strict_train_loss) + 1)

plt.figure(figsize=(9, 5))
plt.plot(strict_rounds, strict_train_loss, label='Train mlogloss (strict)')
plt.plot(strict_rounds, strict_val_loss,   label='Validation mlogloss (strict)')
plt.axvline(strict_model.best_iteration, color='gray', linestyle='--',
            label=f'Best iteration ({strict_model.best_iteration})')
plt.xlabel('Boosting round')
plt.ylabel('Multiclass log loss')
plt.title('Strict Regularization -- Train vs Validation Loss')
plt.legend()
plt.tight_layout()
plt.show()

strict_train_pred = strict_model.predict(X_train)
strict_test_pred  = strict_model.predict(X_test)
strict_train_acc  = accuracy_score(y_train, strict_train_pred)
strict_test_acc   = accuracy_score(y_test, strict_test_pred)
strict_test_kappa = cohen_kappa_score(y_test, strict_test_pred)

strict_loss_gap = (strict_val_loss[strict_model.best_iteration]
                    - strict_train_loss[strict_model.best_iteration])

print("=== Comparison: 6b (loose) tuned model vs 6c (strict) tuned model ===")
print(f"{'':25}{'6b loose':>12}{'6c strict':>12}")
print(f"{'Train accuracy':25}{train_acc*100:11.2f}%{strict_train_acc*100:11.2f}%")
print(f"{'Test accuracy':25}{test_acc*100:11.2f}%{strict_test_acc*100:11.2f}%")
print(f"{'Train/test gap (pts)':25}{(train_acc-test_acc)*100:11.2f} {(strict_train_acc-strict_test_acc)*100:11.2f}")
print(f"{'Test Kappa':25}{test_kappa_tuned:11.4f} {strict_test_kappa:11.4f}")
print(f"{'Val-train loss gap':25}{final_gap:11.4f} {strict_loss_gap:11.4f}")
```

```python
# ---- 6c.4 Choose which model drives the Islamabad classification (Section 9) ----
# Pick the strict model only if it doesn't meaningfully cost test accuracy (within 1 pt)
# -- a smaller train/test and loss gap with similar test performance is a strictly
# more trustworthy model for an unseen deployment area like Islamabad.
if strict_test_acc >= test_acc - 0.01:
    model = strict_model
    print(f"Using the STRICT model (test acc {strict_test_acc*100:.2f}%, "
          f"gap {(strict_train_acc-strict_test_acc)*100:.2f} pts) for downstream classification.")
else:
    model = tuned_model
    print(f"Strict model cost too much test accuracy ({strict_test_acc*100:.2f}% vs "
          f"{test_acc*100:.2f}%) -- keeping the 6b tuned model for downstream classification.")
```

```python
import geopandas as gpd
import pystac_client
import rioxarray  # registers the .rio accessor used below
from odc.stac import configure_s3_access, load


def filter_to_islamabad(gdf, keep_name=KEEP_NAME, drop_names=DROP_NAMES,
                        source_label=AOI_PATH):
    """Keep only the Islamabad feature; drop Chakwal / anything else."""
    if len(gdf) == 0:
        raise ValueError(f'{source_label} has no features.')

    str_cols = [c for c in gdf.columns if c != 'geometry' and gdf[c].dtype == object]

    if len(gdf) == 1:
        if not str_cols:
            print("Single feature with no name attributes; assuming Islamabad.")
            return gdf.copy()
        text = ' '.join(gdf[str_cols].astype(str).values.flatten()).lower()
        if keep_name in text:
            return gdf.copy()
        if any(d in text for d in drop_names):
            raise ValueError(f"Single feature looks like {drop_names}, not '{keep_name}'.")
        print(f"Warning: could not confirm '{keep_name}'; using the single feature found.")
        return gdf.copy()

    keep_mask = None
    for c in str_cols:
        m = gdf[c].astype(str).str.lower().str.contains(keep_name, na=False)
        keep_mask = m if keep_mask is None else (keep_mask | m)
    if keep_mask is not None and keep_mask.any():
        return gdf[keep_mask].copy()

    drop_mask = None
    for c in str_cols:
        col = gdf[c].astype(str).str.lower()
        for d in drop_names:
            m = col.str.contains(d, na=False)
            drop_mask = m if drop_mask is None else (drop_mask | m)
    if drop_mask is not None and drop_mask.any():
        return gdf[~drop_mask].copy()

    print(f"Warning: could not identify '{keep_name}'; using the largest polygon.")
    return gdf.loc[[gdf.geometry.area.idxmax()]].copy()


# ---- Local Dask cluster (speeds up the composite) ----
try:
    from dask.distributed import Client
    client = Client.current()
except ValueError:
    client = Client()
print(client)

# ---- Read + filter the AOI ----
if not os.path.exists(AOI_PATH):
    raise FileNotFoundError(f"Upload your AOI to {AOI_PATH} and run again.")

aoi_gdf = gpd.read_file(AOI_PATH)
if aoi_gdf.crs is None:
    aoi_gdf = aoi_gdf.set_crs('EPSG:4326')
aoi_gdf = filter_to_islamabad(aoi_gdf)
print(f"Using {len(aoi_gdf)} feature(s) for the AOI.")

geometry = (aoi_gdf.geometry.union_all()
            if hasattr(aoi_gdf.geometry, 'union_all')
            else aoi_gdf.geometry.unary_union)
```

```python
# ---- Search the Earth Search STAC for Sentinel-2 scenes ----
catalog = pystac_client.Client.open('https://earth-search.aws.element84.com/v1')
configure_s3_access(aws_unsigned=True)

print("Time range:", time_range)   # set in the form cell above

# simplify AOI to reduce request size
geometry = geometry.simplify(tolerance=0.001, preserve_topology=True)

search = catalog.search(
    collections=['sentinel-2-c1-l2a'],
    intersects=geometry,
    datetime=time_range,
    query={'eo:cloud_cover': {'lt': MAX_CLOUD}},
)
items = search.item_collection()
if len(items) == 0:
    raise RuntimeError("No Sentinel-2 scenes matched. Widen the date range or cloud filter.")


# ---- How many scenes / tiles are available ----
def _tile_id(props):
    if props.get('grid:code'):    return props['grid:code']
    if props.get('s2:mgrs_tile'): return props['s2:mgrs_tile']
    z, lb, gs = (props.get('mgrs:utm_zone'), props.get('mgrs:latitude_band'),
                 props.get('mgrs:grid_square'))
    return f"{z}{lb}{gs}" if (z and lb and gs) else None

tiles = sorted({t for t in (_tile_id(it.properties) for it in items) if t})
dates = sorted({(it.datetime.date().isoformat() if it.datetime
                 else str(it.properties.get('datetime', ''))[:10]) for it in items})
print(f"Scenes (images) found : {len(items)}")
print(f"Unique MGRS tiles     : {len(tiles)}  {tiles}")
print(f"Acquisition dates     : {len(dates)}")

# quick look at scene footprints vs AOI
items_gdf = gpd.GeoDataFrame.from_features(items.to_dict(), crs='EPSG:4326')
fig, ax = plt.subplots(figsize=(5, 5))
items_gdf.plot(ax=ax, facecolor='none', edgecolor='black', alpha=0.5)
aoi_gdf.plot(ax=ax, facecolor='blue', alpha=0.5)
ax.set_title(f'STAC Query Results - {len(items)} scenes, {len(tiles)} tiles')
ax.set_axis_off()
plt.show()
```

```python
import rioxarray  # ensure the .rio accessor is registered
import xarray as xr

# ---- Load spectral bands + SCL (Scene Classification Layer) for cloud masking ----
ds = load(
    items,
    bands=list(BAND_ASSETS.values()) + ['scl'],
    resolution=10,
    bbox=geometry.bounds,
    crs='utm',
    chunks={'x': 1024, 'y': 1024},
    groupby='solar_day',
)
image_crs = ds.rio.crs

# ---- Per-pixel cloud mask from the SCL band ----
# Remove: 0 no-data, 1 saturated/defective, 3 cloud shadow,
#         8 cloud (medium prob), 9 cloud (high prob), 10 thin cirrus
BAD_SCL = [0, 1, 3, 8, 9, 10]
clear = ~ds['scl'].isin(BAD_SCL)

spectral = ds[list(BAND_ASSETS.values())]
spectral = spectral.where(spectral != 0)     # drop fill / nodata
spectral = spectral.where(clear)             # drop clouds, shadows, cirrus
spectral = spectral * 0.0001 - 0.1           # DN -> surface reflectance
# (*0.0001 - 0.1 matches the harmonised Sentinel-2 reflectance used in the training data)

# ---- Reduce the time series to one composite (per the STACK_TYPE form choice) ----
print("Compositing:", STACK_TYPE)
if STACK_TYPE == "Median Stack":
    median_composite = spectral.median(dim='time')
elif STACK_TYPE == "Mean Stack":
    median_composite = spectral.mean(dim='time')
elif STACK_TYPE == "Mode Stack":
    from scipy import stats
    def _mode_along_time(a):
        # a: (..., time) -> reduce the last axis, ignoring NaN
        m = stats.mode(a, axis=-1, nan_policy='omit', keepdims=False)
        return np.asarray(m.mode, dtype='float32')
    median_composite = xr.apply_ufunc(
        _mode_along_time, spectral,
        input_core_dims=[['time']],
        dask='parallelized',
        output_dtypes=['float32'],
    )
else:
    raise ValueError(f"Unknown STACK_TYPE: {STACK_TYPE!r}")

# asset names -> Sentinel-2 band names (B2, B3, ...)
median_composite = median_composite.rename(
    {asset: band for band, asset in BAND_ASSETS.items()})

# ---- Spectral indices (identical formulas to the training data) ----
eps = 1e-6
b = median_composite
median_composite['NDVI'] = (b['B8'] - b['B4'])  / (b['B8'] + b['B4']  + eps)   # (B8-B4)/(B8+B4)
median_composite['NDMI'] = (b['B8'] - b['B11']) / (b['B8'] + b['B11'] + eps)   # (B8-B11)/(B8+B11)
median_composite['NDWI'] = (b['B3'] - b['B8'])  / (b['B3'] + b['B8']  + eps)   # (B3-B8)/(B3+B8)

# 13 bands in the exact model order, then compute
composite = median_composite[FEATURE_ORDER]

from dask.distributed import progress
composite = composite.persist()
progress(composite, notebook=False)
composite = composite.compute()

# stack into a single DataArray, (re)attach CRS, and clip to the AOI
composite_da = composite.to_array('band').assign_coords(band=FEATURE_ORDER)
composite_da = composite_da.rio.write_crs(image_crs)
aoi_reproj = aoi_gdf.to_crs(image_crs)
composite_clipped = composite_da.rio.clip(aoi_reproj.geometry, aoi_reproj.crs)
composite_clipped.rio.write_nodata(np.nan, inplace=True)
print("Stack shape (bands, H, W):", composite_clipped.shape)
```

```python
# ---- Save the 13-band stack to Drive ----
IMG_PATH = f"{OUT_DIR}/{AOI_TAG}_Image.tif"
composite_clipped.rio.to_raster(IMG_PATH, driver='GTiff')

# write band names into the GeoTIFF so they are readable later
with rasterio.open(IMG_PATH, 'r+') as dst:
    dst.descriptions = tuple(FEATURE_ORDER)

print("Saved multiband stack ->", IMG_PATH)
print("Bands:", FEATURE_ORDER)
```

```python
def show_rgb(da, bands, title, stretch=(2, 98)):
    rgb = da.sel(band=bands).values          # (3, H, W)
    rgb = np.moveaxis(rgb, 0, -1)            # (H, W, 3)
    vmin, vmax = np.nanpercentile(rgb, stretch)
    rgb = np.clip((rgb - vmin) / (vmax - vmin), 0, 1)
    plt.figure(figsize=(9, 9))
    plt.imshow(rgb)
    plt.title(title)
    plt.axis('off')
    plt.show()

# True colour: R=B4, G=B3, B=B2
show_rgb(composite_clipped, ['B4', 'B3', 'B2'],
         f"Islamabad - True Colour (B4/B3/B2) {SEASON_LABEL}")

# False colour / colour-infrared: R=B8(NIR), G=B4(Red), B=B3(Green) -> vegetation = red
show_rgb(composite_clipped, ['B8', 'B4', 'B3'],
         f"Islamabad - False Colour / CIR (B8/B4/B3) {SEASON_LABEL}")
```

```python
# ---- Read the saved stack and put bands in the model's feature order ----
with rasterio.open(IMG_PATH) as src:
    arr     = src.read().astype('float32')   # (bands, H, W)
    profile = src.profile.copy()
    nodata  = src.nodata
    desc    = list(src.descriptions)

if desc and all(f in desc for f in FEATURES):
    arr = arr[[desc.index(f) for f in FEATURES]]
assert arr.shape[0] == len(FEATURES), "Band count does not match FEATURES."

bands, H, W = arr.shape
flat = arr.reshape(bands, -1).T              # (pixels, bands)

# valid pixels = finite in every band
valid = np.all(np.isfinite(flat), axis=1)
if nodata is not None and np.isfinite(nodata):
    valid &= ~np.any(flat == nodata, axis=1)

# predict (0 reserved for nodata)
pred = np.zeros(flat.shape[0], dtype=np.uint8)
X_pred = pd.DataFrame(flat[valid], columns=FEATURES)
pred[valid] = le.inverse_transform(model.predict(X_pred)).astype(np.uint8)
classification = pred.reshape(H, W)

# ---- Save classification.tif ----
out_profile = profile.copy()
out_profile.update(count=1, dtype='uint8', nodata=0, driver='GTiff')
for _k in ('blockxsize', 'blockysize', 'interleave', 'photometric',
           'tiled', 'compress', 'predictor'):
    out_profile.pop(_k, None)

CLS_PATH = f"{OUT_DIR}/{AOI_TAG}_classification.tif"
with rasterio.open(CLS_PATH, 'w', **out_profile) as dst:
    dst.write(classification, 1)
print("Saved classification ->", CLS_PATH)
```

```python
# Colour for each class number (1-12)
class_colors = {
    1:  '#c8c8c8',  # Bare_BuiltUp
    2:  '#ffd400',  # Corn
    3:  '#ff69b4',  # Cotton
    4:  '#aadd77',  # Grass_Pasture
    5:  '#ff8c00',  # Jowar_Sorghum
    6:  '#a0522d',  # Pulses
    7:  '#00e5ff',  # Rice
    8:  '#9acd32',  # Soybeans
    9:  '#006400',  # Sugarcane
    10: '#ffff33',  # Sunflower
    11: '#228b22',  # Tree_Cover
    12: '#1f4fff',  # Water
}

def plot_classes(arr2d, title):
    ids    = sorted(id_to_name)                              # [1..12]
    colors = ['#000000'] + [class_colors[i] for i in ids]   # 0 = nodata (black)
    cmap   = ListedColormap(colors)
    norm   = BoundaryNorm(np.arange(-0.5, len(ids) + 1, 1), cmap.N)

    plt.figure(figsize=(11, 9))
    plt.imshow(arr2d, cmap=cmap, norm=norm)
    plt.title(title)
    plt.axis('off')
    patches = [mpatches.Patch(color=class_colors[i], label=f"{i} — {id_to_name[i]}")
               for i in ids]
    plt.legend(handles=patches, bbox_to_anchor=(1.02, 1), loc='upper left',
               borderaxespad=0., title="Class")
    plt.tight_layout()
    plt.show()

plot_classes(classification, "Islamabad — XGBoost Classification")
```

```python
from skimage.filters.rank import majority
from skimage.morphology import square

WINDOW = 3                                     # 3x3 neighbourhood
cleaned = majority(classification, square(WINDOW))
cleaned[classification == 0] = 0               # keep original nodata as nodata

# ---- Save classification_cleaned.tif ----
CLEAN_PATH = f"{OUT_DIR}/{AOI_TAG}_classification_cleaned.tif"
with rasterio.open(CLEAN_PATH, 'w', **out_profile) as dst:
    dst.write(cleaned.astype(np.uint8), 1)
print("Saved cleaned classification ->", CLEAN_PATH)

plot_classes(cleaned, f"Islamabad — Majority Filtered ({WINDOW}x{WINDOW})")
```
